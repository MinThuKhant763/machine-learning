```python
import numpy as np
import pandas as pd
x = np.arange(0,20,0.4)
print(x)
y1 = np.sin(x)
print(y1)
y2 = np.cos(x)*50
print(y2)
```

```python
import matplotlib.pyplot as plt
plt.plot(x,y1,'red')
plt.plot(x,y2,'blue')
plt.show()
```

```python
y1_new = y1/max(y1)
y2_new = y2/max(y2)
plt.plot(x,y1_new,'red')
plt.plot(x,y2_new,'blue')
```

```python
y1_new = (y1-min(y1))/(max(y1)-min(y1))
y2_new = (y2-min(y2))/(max(y2)-min(y2))
plt.plot(x,y1_new,"red")
plt.plot(x,y2_new,"blue")
```

```python
from scipy import stats
IQR1 = stats.iqr(y1,interpolation="midpoint")
y1_new = (y1-np.mean(y1))/IQR1

IQR2 = stats.iqr(y2,interpolation="midpoint")
y2_new = (y2-np.mean(y2))/IQR2

plt.plot(x,y1_new,'red')
plt.plot(x,y2_new,'blue')
plt.show()
```
