```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
data = {"Size(sqrt)": [1500,1600,2400,1400,3000],
        "Bedrooms": [3,3,4,2,4],
        "Age(years)": [10,15,20,5,8],
        "Price($)": [400000,420000,600000,350000,650000]}
df = pd.DataFrame(data)
print(df)
```

```python
X = df[["Size(sqrt)","Bedrooms","Age(years)"]]
Y = df["Price($)"]
print(X)
print(Y)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
print("Train Features:\n",X_train)
print("Test Feature:\n",X_test)
print("Train Target:\n",Y_train)
print("Test Target:\n",Y_test)
```

```python
model = LinearRegression()
model.fit(X_train,Y_train)
Y_pred = model.predict(X_test)
print("Predicteed Prices ",Y_pred)
```

```python
mse = mean_squared_error(Y_test,Y_pred)
print("Mean Square Error : ",mse)
#r2 = r2_score(Y_test,Y_pred)
#print("R2 Score : ", r2)
```

```python
new_house = [[2000,3,10]]
predicited_price = model.predict(new_house)
print("Predicit Price for new House : $",round(predicited_price[0],2))
```
